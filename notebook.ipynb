```python
import numpy as np
from cvrp_encoding import (
    CVRPInstance,
    decode_two_part,
    decode_giant_tour_greedy_split,
    mo_objectives,
    total_distance,
)

# Course example (1-based in statement) -> 0-based genes
assign = [0, 1, 0, 0, 1]  # was [1,2,1,1,2]
order_perm = [2, 0, 4, 1, 3]  # was [3,1,5,2,4] — global visit order
routes_tp = decode_two_part(assign, order_perm, n_vehicles=2)
print("Strategy 1 routes (customer idx 0..N-1):", routes_tp)

rng = np.random.default_rng(0)
inst = CVRPInstance(
    depot_xy=np.zeros(2),
    customer_xy=rng.random((5, 2)) * 10.0,
    demands=np.ones(5),
    vehicle_capacity=3.0,
    n_vehicles=2,
)
f1, f2 = mo_objectives(inst, routes_tp, capacity_penalty_weight=0.0)
print("distance", f1, "load_balance_std", f2)

routes_gt = decode_giant_tour_greedy_split(order_perm, inst)
print("Strategy 2 (same permutation, greedy split):", routes_gt)
print("objectives", mo_objectives(inst, routes_gt))
```
